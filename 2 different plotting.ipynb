```python
import pandas as pd

url = "https://raw.githubusercontent.com/reemajedz99/IT326-Diabetes/refs/heads/main/Dataset/Raw_dataset.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.info()
df.describe()
df.columns
```

```python
import matplotlib.pyplot as plt

plt.hist(df["BMI"], bins=10)
plt.xlabel("BMI")
plt.ylabel("Frequency")
plt.title("Distribution of BMI")
plt.show()
```

**Description**:

Displays the frequency distribution of Body Mass Index (BMI) values across all patients, spanning from $15$ to $40$.  

**Key Info Provided**:

 Shows a uniform distribution with roughly equal counts across all BMI ranges, lacking any distinct peak or skewness.  
  **Why Preprocessing is Needed**:
  
  Feature Binning / Discretization: The uniform spread without natural clustering justifies discretizing continuous BMI values into standard clinical categories (Underweight, Normal, Overweight, Obese) to improve model interpretability.Feature Scaling: Uniform distribution across a wide range ($15\text{--}40$) highlights the need for normalization/standardization to align feature scales across the dataset

```python
plt.scatter(df["Age"], df["BMI"])
plt.xlabel("Age")
plt.ylabel("BMI")
plt.title("Relationship between Age and BMI")
plt.show()
```

**Description**:

Illustrates the two-dimensional relationship between patient Age (X-axis) and BMI (Y-axis).

   **Key Info Provided**:
   
   Displays an evenly scattered rectangular cloud of data points, indicating no linear correlation (zero relationship) between Age and BMI.

   **Why Preprocessing is Needed**:

   Feature Selection & Reduction: The complete independence between Age and BMI reveals that they lack joint predictive power, justifying feature ranking or dropping weak features in favor of highly predictive medical attributes (like FastingBloodSugar).   Feature Scaling (Normalization): Shows two features operating on different numerical ranges (Age: $20\text{--}90$ vs. BMI: $15\text{--}40$). Preprocessing via MinMaxScaler or StandardScaler is required to prevent distance-based algorithms (e.g., KNN, SVM) from being biased toward Age.
