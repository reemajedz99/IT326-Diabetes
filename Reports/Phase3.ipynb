# [4] Data Preprocessing

## 1. Normalization
The code normalizes the data by applying MinMaxScaler from sklearn.preprocessing. It scales all 12 continuous numeric columns in the dataset so that their values are transformed into a range between 0 and 1, while keeping the target column (Diagnosis) unchanged.

**Justification:**

The columns in our dataset use different units. For example, total cholesterol has much larger numerical values than sleep quality. We used normalization so we could compare patient records on a common scale.

**Why was Normalization applied:**

Without normalization, a difference in cholesterol could have a much larger effect on the distance between two patients just because of the units used to measure it.

**How was Normalization applied:**

We used MinMaxScaler from sklearn.preprocessing to scale each selected column to a range from 0 to 1. The smallest value became 0, the largest became 1, and the values in between were scaled proportionally.

**Applied Attributes:**

AlcoholConsumption, PhysicalActivity, DietQuality, SleepQuality, SerumCreatinine, BUNLevels, CholesterolTotal, CholesterolLDL, CholesterolHDL, CholesterolTriglycerides, FatigueLevels, and QualityOfLifeScore.

**How did Normalization improve the dataset:**

After normalization, each selected column had the same range. This made the data more suitable for clustering because differences between patients could be compared across columns without larger units dominating the calculation.

```python
import os
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

raw_df = pd.read_csv('https://raw.githubusercontent.com/reemajedz99/IT326-Diabetes/main/Dataset/Raw_dataset.csv')

preprocessed_df = raw_df.copy()

num_cols = [
    'AlcoholConsumption',
    'PhysicalActivity',
    'DietQuality',
    'SleepQuality',
    'SerumCreatinine',
    'BUNLevels',
    'CholesterolTotal',
    'CholesterolLDL',
    'CholesterolHDL',
    'CholesterolTriglycerides',
    'FatigueLevels',
    'QualityOfLifeScore',
]

scaler = MinMaxScaler()
preprocessed_df[num_cols] = scaler.fit_transform(preprocessed_df[num_cols])

print("Raw Dataset Sample:")
display(raw_df[num_cols].head())

print("\nNormalized Dataset Sample:")
display(preprocessed_df[num_cols].head())
```

## 2. Discretization


```python
import numpy as np
import pandas as pd

clinical = {
    'Age': (
        [-np.inf, 40, 60, np.inf],
        ['Young adult', 'Middle-aged', 'Senior'],
),
    'BMI': (
        [-np.inf, 18.5, 25, 30, np.inf],
        ['Underweight', 'Normal', 'Overweight', 'Obese'],
    ),
    'HbA1c': ([-np.inf, 5.7, 6.5, np.inf], ['Normal', 'Prediabetes', 'Diabetes']),
    'FastingBloodSugar': (
        [-np.inf, 100, 126, np.inf],
        ['Normal', 'Prediabetes', 'Diabetes'],
    ),
    'SystolicBP': (
        [-np.inf, 120, 130, 140, np.inf],
        ['Normal', 'Elevated', 'Stage 1', 'Stage 2'],
    ),
    'DiastolicBP': (
        [-np.inf, 80, 90, np.inf],
        ['Normal', 'Stage 1', 'Stage 2'],
    ),
}

binned = pd.DataFrame({
    c: pd.cut(raw_df[c], bins=e, labels=l, right=False)
    for c, (e, l) in clinical.items()
})

for col in clinical:
  preprocessed_df[col] = binned[col]

print("Raw sample:")
display(raw_df[list(clinical)].head())

print("\nDiscretized sample:")
display(preprocessed_df[list(clinical)].head())
```

**Justification:**

In our diabetes dataset, features like blood sugar, blood pressure, and BMI directly correlate with diagnostic stages. Grouping raw patient measurements into tiers (such as 'Normal', 'Prediabetes', and 'Diabetes') turns continuous values into interpretable groups, improving data quality and making extracted patterns human-readable for medical analysis.

**Why was Discretizing applied:**

Exact numerical health measurements can contain small variations and minor differences that make classification models overly complex. Discretizing these continuous values into domain-based categories ensures results strictly follow real known medical standards rather than arbitrary mathematical splits.  

**How was Discretizing applied:**

Left-inclusive interval mapping (pd.cut(..., right=False)) was applied directly to unscaled values from raw_df using standard medical thresholds.

**On which attributes:**

Age, BMI, HbA1c, FastingBloodSugar, SystolicBP, and DiastolicBP.

**How did Discretizing improve the dataset:**
The resulting dataset is more interpretable, as continuous clinical measurements are represented as meaningful categories that make patient patterns and risk levels easier to identify.

# [5] Data Mining Technique
##Decision Tree Classification:

Decision Tree Classification is a supervised learning algorithm used for classification and regression tasks, predicting categorical outcomes based on the input features. In this report, this classification technique is applied to predict diabetes daignosis using patient health information (age, BMI, blood sugar levels, lifestyle habits etc).


#The attribute measures that have been used are:

**1. Gini Index**

Gini Index measures the impurity of the data at each node. The attribute that provides the **largest impurity reduction** is selected as the splitting attribute. It helps the decision tree separate patients into different diabetes diagnosis classes. The **DecisionTreeClassifier(criterion='gini')** method calculates Gini impurity and selects the split that produces the greatest reduction in impurity at each node.


**2. Information Gain (Entropy)**

Information Gain or Entropy measures the uncertainty after splitting the data. The attribute with the **highest information gain** is chosen as the splitting attribute. It helps in identifying the attributes that provide the most information about diabetes diagnosis. The **DecisionTreeClassifier(criterion='entropy')** method calculates the information gain for possible splits and selects the split with the highest information gain at each node.

# [6] Evaluation and Comparison
## Classification: 70% Training and 30% Testing




This code separates Diagnosis from the other patient attributes. The model uses both during training, but uses only the other attributes to predict Diagnosis during testing. It uses 70% of the records for training and 30% for testing. Using stratify=y keeps the percentage of patients with and without diabetes approximately the same in both the training and testing sets as in the full dataset. Using random_state=42 ensures that the same records are selected for training and testing each time the code runs on the same dataset.



```python
from sklearn.model_selection import train_test_split

X = encoded_df.drop(columns=['Diagnosis'])
y = encoded_df['Diagnosis']

X_train, X_test, y_train, y_test = train_test_split(
X,
y,
test_size=0.30,
random_state=42,
stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

This code trains a decision tree on the training data using Gini to choose how to split the data within the tree. The model then predicts the class labels (Diagnosis: 0 or 1) for the test records. It compares these predictions with the actual class labels to calculate accuracy, the percentage of correct predictions. Finally, it displays a confusion matrix showing how many records from each class were predicted correctly or incorrectly.



```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

gini_model = DecisionTreeClassifier(
criterion='gini',
random_state=42
)

gini_model.fit(X_train, y_train)

gini_predictions = gini_model.predict(X_test)

gini_accuracy = accuracy_score(y_test, gini_predictions)

print(f"Gini Accuracy: {gini_accuracy:.4f}")
print(f"Gini Accuracy (%): {gini_accuracy * 100:.2f}%")

ConfusionMatrixDisplay.from_predictions(
y_test,
gini_predictions,
labels=[0, 1],
display_labels=['No Diabetes', 'Diabetes'],
cmap='Blues'
)

plt.title('Gini Confusion Matrix — 70/30 Split')
plt.show()
```

This code trains a decision tree on the same training data used for Gini, using entropy to choose how to split the data within the tree. The model then predicts the class labels (Diagnosis: 0 or 1) for the test records. It compares these predictions with the actual class labels to calculate accuracy, the percentage of correct predictions. Finally, it displays a confusion matrix showing how many records from each class were predicted correctly or incorrectly. Both models use the same training and testing records to make the comparison fair.



```python
entropy_model = DecisionTreeClassifier(
criterion='entropy',
random_state=42
)

entropy_model.fit(X_train, y_train)

entropy_predictions = entropy_model.predict(X_test)

entropy_accuracy = accuracy_score(y_test, entropy_predictions)

print(f"Entropy Accuracy: {entropy_accuracy:.4f}")
print(f"Entropy Accuracy (%): {entropy_accuracy * 100:.2f}%")

ConfusionMatrixDisplay.from_predictions(
y_test,
entropy_predictions,
labels=[0, 1],
display_labels=['No Diabetes', 'Diabetes'],
cmap='Greens'
)

plt.title('Entropy Confusion Matrix — 70/30 Split')
plt.show()
```

This code displays the accuracy percentages of Gini and entropy in a table to compare their results on the 70/30 split.



```python
results_70_30 = pd.DataFrame({
'Criterion': ['Gini', 'Entropy'],
'Training (%)': [70, 70],
'Testing (%)': [30, 30],
'Accuracy (%)': [
gini_accuracy * 100,
entropy_accuracy * 100
]
})

display(results_70_30.round(2))
```

Both decision tree models were evaluated using the same 70/30 split, with 1,315 training records and 564 test records. Diagnosis = 1 represents diabetes, and Diagnosis = 0 represents no diabetes.

The Gini model achieved 88.48% accuracy, correctly classifying 499 of the 564 test records. # **1- Justification for Choice of K Values.**

### K-means is applied without the class label (Diagnosis) since clustering is unsupervised:

K = 2: matches the two classes of the label (diabetes / no diabetes)

K = 3: low / moderate / high risk profiles ,at the pre-diagnostic stage before diabetes is confirmed.

K = 4: finer groups that may combine metabolic factors and lifestyle habits.

The Elbow and Silhouette curves for K = 2 to 10 are used to confirm that these values are reasonable.


```python
import pandas as pd

url = "https://raw.githubusercontent.com/reemajedz99/IT326-Diabetes/refs/heads/main/Dataset/Preprocessed_dataset.csv"
encoded_df = pd.read_csv(url)

print("Dataset shape:", encoded_df.shape)
print("Text columns:", encoded_df.select_dtypes(include="object").columns.tolist())
```

This cell loads Preprocessed_dataset.csv from the project repository into encoded_df.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.decomposition import PCA

# Drop class label 'Diagnosis' if present
X_cluster = encoded_df.drop(columns=['Diagnosis', 'PatientID' , 'DoctorInCharge'], errors='ignore')
X_cluster = X_cluster.select_dtypes(include='number')
X_scaled = MinMaxScaler().fit_transform(X_cluster)
print(X_scaled.shape)
```

The class label (Diagnosis) and the identifier columns are removed

```python
k_range = range(2, 11)
wss_list, sil_list = [], []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    wss_list.append(km.inertia_)
    sil_list.append(silhouette_score(X_scaled, labels))

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

ax[0].plot(list(k_range), wss_list, marker='o')
ax[0].set_title('Elbow Method (Total within-cluster SS)')
ax[0].set_xlabel('K'); ax[0].set_ylabel('WSS')

ax[1].plot(list(k_range), sil_list, marker='o', color='green')
ax[1].set_title('Average Silhouette Width')
ax[1].set_xlabel('K'); ax[1].set_ylabel('Silhouette')

for a in ax:
    for k in (2, 3, 4):
        a.axvline(k, color='red', linestyle='--', alpha=0.3)

plt.tight_layout(); plt.show()
```

Elbow and Silhouette curves for K = 2 to 10
K-means was run for K = 2 to 10, and the total within-cluster sum of squares (WSS) and the average Silhouette width were recorded for each K.

```python
chosen_k = [2, 3, 4]
models, results = {}, []

for k in chosen_k:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    models[k] = (km, labels)
    results.append({
        'K': k,
        ' Silhouette ': silhouette_score(X_scaled, labels),
        'Total within-cluster SS (WSS)': km.inertia_,
        'Davies-Bouldin (lower=better)': davies_bouldin_score(X_scaled, labels),
        'Cluster sizes': sorted(np.bincount(labels).tolist(), reverse=True)
    })

results_df = pd.DataFrame(results).set_index('K')
display(results_df.round(4))
```

:Silhouette is highest for K = 4 (0.0468) and K = 2 (0.0467), and Davies-Bouldin is lowest for K = 3 (4.2357).

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
print('Variance explained by PC1 + PC2: {:.1f}%'.format(pca.explained_variance_ratio_.sum() * 100))

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, k in zip(axes, chosen_k):
    km, labels = models[k]
    centers_pca = pca.transform(km.cluster_centers_)
    ax.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='tab10', s=15, alpha=0.6)
    ax.scatter(centers_pca[:, 0], centers_pca[:, 1], c='black', marker='X',
               s=200, edgecolors='white', label='Centroids')
    ax.set_title(f'K = {k}')
    ax.set_xlabel('PC1'); ax.set_ylabel('PC2'); ax.legend()
plt.suptitle('K-means clusters projected on the first two principal components')
plt.tight_layout(); plt.show()
```

Cluster visualization
The 44 features were projected to two dimensions using PCA so the clusters can be plotted. Each color is a cluster and each black X is a centroid. The first two components explain only 9.4%

```python
from sklearn.metrics import silhouette_samples
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, k in zip(axes, chosen_k):
    _, labels = models[k]
    sil_vals = silhouette_samples(X_scaled, labels)
    y_lower = 10
    for c in range(k):
        v = np.sort(sil_vals[labels == c])
        y_upper = y_lower + len(v)
        ax.fill_betweenx(np.arange(y_lower, y_upper), 0, v, alpha=0.7, label=f'Cluster {c}')
        y_lower = y_upper + 10
    ax.axvline(sil_vals.mean(), color='red', linestyle='--', label='Average')
    ax.set_title(f'K = {k}  (avg = {sil_vals.mean():.4f})')
    ax.set_xlabel('Silhouette coefficient'); ax.set_yticks([]); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
```

Silhouette plot for each K
 All clusters have mostly small positive values and some records fall below zero, which confirms that the clusters overlap and are weakly separated.

```python
ks = np.array(list(k_range), dtype=float)
w = np.array(wss_list)
xn = (ks - ks.min()) / (ks.max() - ks.min())
yn = (w - w.min()) / (w.max() - w.min())
knee_dist = dict(zip(k_range, np.abs(yn - (1 - xn)) / np.sqrt(2)))

k_values = chosen_k
sil = results_df[' Silhouette '].to_dict()
dbi = results_df['Davies-Bouldin (lower=better)'].to_dict()

votes = {
    'Silhouette (max)': max(k_values, key=lambda k: sil[k]),
    'Elbow / WSS (knee)': max(k_values, key=lambda k: knee_dist[k]),
    'Davies-Bouldin (min)': min(k_values, key=lambda k: dbi[k]),
}
for name, v in votes.items():
    print(f"{name:22s} -> K = {v}")

best_k = Counter(votes.values()).most_common(1)[0][0]
print("Best K by majority rule: K =", best_k)
```

```python
# Extract Silhouette and WSS metrics from the results_df DataFrame
sil = results_df[' Silhouette '].to_dict()
wss = results_df['Total within-cluster SS (WSS)'].to_dict()

results_clustering = pd.DataFrame({
    f'K={k}': {
        'Average Silhouette width': round(sil[k], 4),
        'Total within-cluster sum of square': round(wss[k], 2),
    } for k in k_values
})
display(results_clustering)
```

The table summarizes the Average Silhouette width and the total within-cluster sum of square for each K. K = 4 has the highest Silhouette (0.0468) and the lowest WSS (9164.39).

##Classification

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree

url = "https://raw.githubusercontent.com/reemajedz99/IT326-Diabetes/refs/heads/main/Dataset/Raw_dataset.csv"
encoded_df = pd.read_csv(url)

encoded_df = encoded_df.drop(columns=['PatientID', 'DoctorInCharge'], errors='ignore')

X = encoded_df.drop(columns=['Diagnosis'])
y = encoded_df['Diagnosis']

X_train_90, X_test_90, y_train_90, y_test_90 = train_test_split(
    X, y, test_size=0.10, random_state=42, stratify=y
)

entropy_model_90 = DecisionTreeClassifier(criterion='entropy', random_state=42)
entropy_model_90.fit(X_train_90, y_train_90)

best_model = entropy_model_90
feature_names = X_train_90.columns.tolist()

plt.figure(figsize=(24, 12))
plot_tree(best_model,
          feature_names=feature_names,
          class_names=['No Diabetes', 'Diabetes'],
          filled=True,
          rounded=True,
          max_depth=4,
          fontsize=10)
plt.title("Final Decision Tree (Entropy, 90/10 split) — Top 4 levels", fontsize=18)
plt.show()

importances = pd.Series(best_model.feature_importances_, index=feature_names).sort_values(ascending=False)
plt.figure(figsize=(10, 6))
importances.head(15).plot(kind='barh', color='#2a6f8e')
plt.gca().invert_yaxis()
plt.title("Top 15 Feature Importances (Entropy Model)", fontsize=14)
plt.xlabel("Importance Score")
plt.show()
```

### The Final Decision Tree

The visual above shows the top four levels of our best-performing model (Entropy criterion, 90/10 split).

*   **The Root Node:** The tree starts with **FastingBloodSugar**. The algorithm found that asking "Is the patient's fasting blood sugar <= 126.7?" provides the highest Information Gain. Notice how the left branch (True) is mostly blue (No Diabetes), while the right branch (False) is mostly orange (Diabetes).
*   **Next Splits:** On the right branch (patients with high blood sugar), the tree immediately asks about **HbA1c**. If HbA1c is also high, the patient is classified as having Diabetes. On the left branch, the tree uses symptoms like **FrequentUrination** to catch any diabetic patients who might have slipped through with lower fasting blood sugar.
*   **Color Code:** The boxes are colored based on the majority class in that node. Blue represents "No Diabetes" (Class 0) and Orange represents "Diabetes" (Class 1).
*   **Numbers:** Inside each box, `samples` shows the total number of patients that reached that point, and `value` shows the exact count of `[No Diabetes, Diabetes]`.
*   **Limit:** We limited this plot to a `max_depth` of 4 so it is readable. The actual tree is much deeper and continues splitting until the boxes are 100% pure. This deep splitting is exactly why our model overfitted and struggled on the test data.
*   **Feature Importance:** The bar chart confirms the tree's structure. While `FastingBloodSugar` is the root split, **HbA1c** actually has the highest overall importance score because the tree uses it repeatedly deeper in the branches to separate the classes. `FrequentUrination` and `Hypertension` are the next most useful features.

This project aimed to analyze the diabetes health dataset from two perspectives: predicting the diagnosis (Classification) and discovering natural patient profiles (Clustering).

As for classification, we evaluated Decision Trees using Gini and Entropy across three train/test splits.

The best classification model was the **Entropy Decision Tree (90/10 split)** achieving **59.04% accuracy**. However, because the dataset is imbalanced (60% No Diabetes vs. 40% Diabetes), a baseline model guessing "No Diabetes" every time would achieve 60%. Our model did not clearly beat this baseline.
*   **Visual Representation & Insights:** As seen in the tree plot above, the model's root splits are on **HbA1c** and **FastingBloodSugar**. This is a highly meaningful finding: the algorithm successfully rediscovered the standard clinical definition of diabetes.
