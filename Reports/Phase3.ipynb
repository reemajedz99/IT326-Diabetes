# [4] Data Preprocessing

## 1. Normalization

The code normalizes the data by applying `MinMaxScaler` from `sklearn.preprocessing`. It scales all 12 continuous numeric columns in the dataset so that their values are transformed into a range between `0` and `1`, while keeping the target column (`Diagnosis`) unchanged.

#### **Applied Attributes:**
`AlcoholConsumption`, `PhysicalActivity`, `DietQuality`, `SleepQuality`, `SerumCreatinine`, `BUNLevels`, `CholesterolTotal`, `CholesterolLDL`, `CholesterolHDL`, `CholesterolTriglycerides`, `FatigueLevels`, and `QualityOfLifeScore`.

#### **Justification:**
In our raw dataset, numerical features have very different ranges. Scaling the data ensures that all features are uniform, preventing large numbers from dominating distance calculations and improving overall model performance.
####**How did Normalization improve the dataset:**

The resulting dataset provides balanced, uniformly scaled inputs that prevent feature bias, improving overall convergence and accuracy for subsequent machine learning models.

```python
import os
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

raw_df = pd.read_csv('Raw_dataset.csv')

preprocessed_df = raw_df.copy()

num_cols = [
    'AlcoholConsumption',
    'PhysicalActivity',
    'DietQuality',
    'SleepQuality',
    'SerumCreatinine',
    'BUNLevels',
    'CholesterolTotal',
    'CholesterolLDL',
    'CholesterolHDL',
    'CholesterolTriglycerides',
    'FatigueLevels',
    'QualityOfLifeScore',
]

scaler = MinMaxScaler()
preprocessed_df[num_cols] = scaler.fit_transform(preprocessed_df[num_cols])

print("Raw Dataset Sample:")
display(raw_df[num_cols].head())

print("\nNormalized Dataset Sample:")
display(preprocessed_df[num_cols].head())
```

## 2. Discretization


```python
import numpy as np
import pandas as pd

clinical = {
    'Age': (
        [-np.inf, 40, 60, np.inf],
        ['Young adult', 'Middle-aged', 'Senior'],
),
    'BMI': (
        [-np.inf, 18.5, 25, 30, np.inf],
        ['Underweight', 'Normal', 'Overweight', 'Obese'],
    ),
    'HbA1c': ([-np.inf, 5.7, 6.5, np.inf], ['Normal', 'Prediabetes', 'Diabetes']),
    'FastingBloodSugar': (
        [-np.inf, 100, 126, np.inf],
        ['Normal', 'Prediabetes', 'Diabetes'],
    ),
    'SystolicBP': (
        [-np.inf, 120, 130, 140, np.inf],
        ['Normal', 'Elevated', 'Stage 1', 'Stage 2'],
    ),
    'DiastolicBP': (
        [-np.inf, 80, 90, np.inf],
        ['Normal', 'Stage 1', 'Stage 2'],
    ),
}

binned = pd.DataFrame({
    c: pd.cut(raw_df[c], bins=e, labels=l, right=False)
    for c, (e, l) in clinical.items()
})

for col in clinical:
  preprocessed_df[col] = binned[col]

print("Raw sample:")
display(raw_df[list(clinical)].head())

print("\nDiscretized sample:")
display(preprocessed_df[list(clinical)].head())
```

**Justification:**

In our diabetes dataset, features like blood sugar, blood pressure, and BMI directly correlate with diagnostic stages. Grouping raw patient measurements into tiers (such as 'Normal', 'Prediabetes', and 'Diabetes') turns continuous values into interpretable groups. Improving data quality and making extracted patterns human-readable for medical analysis.

**Why was Discretizing applied:**

Exact numerical health measurements can contain small variations and minor differences that make classification models overly complex. Discretizing these continuous values into domain-based categories ensures results strictly follow real known medical standards rather than arbitrary mathematical splits.  

**How was Discretizing applied:**

Left-inclusive interval mapping (pd.cut(..., right=False)) was applied directly to unscaled values from df using standard medical thresholds. After binning.

**On which attributes:**

Age, BMI, HbA1c, FastingBloodSugar, SystolicBP, and DiastolicBP.

**How did Discretizing improve the dataset:**
The resulting dataset is more interpretable, as continuous clinical measurements are represented as meaningful categories that make patient patterns and risk levels easier to identify.

# [5] Data Mining Technique
##Decision Tree Classification:

Decision Tree Classification is a supervised learning algorithm used for classification and regression tasks, predictiing categorical outcomes based on the input features. In this report, this classification technique is applied to predict diabetes daignosis using patient health information (age, BMI, blood sugar levels, lifestyle habits etc).


#The attribute measures that have been used are:

**1. Gini Index**

Gini Index measures the impurity of the data at each node. The attribute that provides the **largest impurity reduction** is selected as the splitting attribute. It helps the decision tree separate patients into different diabetes diagnosis classes. The **DecisionTreeClassifier(criterion='gini')** method calculates Gini impurity and selects the split that produces the greatest reduction in impurity at each node.


**2. Information Gain (Entropy)**

Information Gain or Entropy measures the uncertainty after splitting the data. The attribute with the **highest information gain** is chosen as the splitting attribute. It helps in identifying the attributes that provide the most information about diabetes diagnosis. The **DecisionTreeClassifier(criterion='entropy')** method calculates the information gain for possible splits and selects the split with the highest information gain at each node.

## Classification: 70% Training and 30% Testing




```python
from sklearn.model_selection import train_test_split

X = encoded_df.drop(columns=['Diagnosis'])
y = encoded_df['Diagnosis']

X_train, X_test, y_train, y_test = train_test_split(
X,
y,
test_size=0.30,
random_state=42,
stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

gini_model = DecisionTreeClassifier(
criterion='gini',
random_state=42
)

gini_model.fit(X_train, y_train)

gini_predictions = gini_model.predict(X_test)

gini_accuracy = accuracy_score(y_test, gini_predictions)

print(f"Gini Accuracy: {gini_accuracy:.4f}")
print(f"Gini Accuracy (%): {gini_accuracy * 100:.2f}%")

ConfusionMatrixDisplay.from_predictions(
y_test,
gini_predictions,
labels=[0, 1],
display_labels=['No Diabetes', 'Diabetes'],
cmap='Blues'
)

plt.title('Gini Confusion Matrix — 70/30 Split')
plt.show()
```

```python
entropy_model = DecisionTreeClassifier(
criterion='entropy',
random_state=42
)

entropy_model.fit(X_train, y_train)

entropy_predictions = entropy_model.predict(X_test)

entropy_accuracy = accuracy_score(y_test, entropy_predictions)

print(f"Entropy Accuracy: {entropy_accuracy:.4f}")
print(f"Entropy Accuracy (%): {entropy_accuracy * 100:.2f}%")

ConfusionMatrixDisplay.from_predictions(
y_test,
entropy_predictions,
labels=[0, 1],
display_labels=['No Diabetes', 'Diabetes'],
cmap='Greens'
)

plt.title('Entropy Confusion Matrix — 70/30 Split')
plt.show()
```

```python
results_70_30 = pd.DataFrame({
'Criterion': ['Gini', 'Entropy'],
'Training (%)': [70, 70],
'Testing (%)': [30, 30],
'Accuracy (%)': [
gini_accuracy * 100,
entropy_accuracy * 100
]
})

display(results_70_30.round(2))
```

```python
gini_acc_70, entropy_acc_70 = gini_accuracy , entropy_accuracy
gini_acc_80, entropy_acc_80 = gini_accuracy_80, entropy_accuracy_80
gini_acc_90, entropy_acc_90 = gini_accuracy_90, entropy_accuracy_90

table_gini = pd.DataFrame({
    'Train/Test Split': ['70/30', '80/20', '90/10'],
    'Gini Accuracy (%)': [gini_acc_70 * 100, gini_acc_80 * 100, gini_acc_90 * 100]
})

table_entropy = pd.DataFrame({
    'Train/Test Split': ['70/30', '80/20', '90/10'],
    'Entropy Accuracy (%)': [entropy_acc_70 * 100, entropy_acc_80 * 100, entropy_acc_90 * 100]
})

print("=== Table 1: Gini Index Accuracy Across Splits ===")
display(table_gini.round(2))

print("\n=== Table 2: Entropy Accuracy Across Splits ===")
display(table_entropy.round(2))
```

```python
combined_summary = pd.DataFrame({
    'Train/Test Split': table_gini['Train/Test Split'],
    'Gini Accuracy (%)': table_gini['Gini Accuracy (%)'],
    'Entropy Accuracy (%)': table_entropy['Entropy Accuracy (%)']
})

gap = combined_summary['Gini Accuracy (%)'] - combined_summary['Entropy Accuracy (%)']
combined_summary['Best Algorithm for Split'] = np.where(
    np.isclose(gap, 0), 'Tie', np.where(gap > 0, 'Gini', 'Entropy')
)

print("=== Comparison & Best Algorithm for Each Split ===")
display(combined_summary.round(2))
```

```python
accuracy_columns = ['Gini Accuracy (%)', 'Entropy Accuracy (%)']
best_value = combined_summary[accuracy_columns].max().max()

best_matches = []
for _, row in combined_summary.iterrows():
    for criterion in accuracy_columns:
        if np.isclose(row[criterion], best_value):
            best_matches.append({
                'Criterion': criterion.replace(' Accuracy (%)', ''),
                'Train/Test Split': row['Train/Test Split'],
                'Accuracy (%)': best_value
            })

print("=== Overall Best Model ===")
display(pd.DataFrame(best_matches))
```

# **1- Justification for Choice of K Values.**

K = 2: matches the two classes of the label (diabetes / no diabetes)

K = 3: low / moderate / high risk profiles ,at the pre-diagnostic stage before diabetes is confirmed.

K = 4: finer groups that may combine metabolic factors and lifestyle habits.

```python
import pandas as pd
import io
from google.colab import files

uploaded = files.upload()
filename = next(iter(uploaded))

encoded_df = pd.read_csv(io.BytesIO(uploaded[filename]))

print("Dataset shape:", encoded_df.shape)
print("Text columns:", encoded_df.select_dtypes(include="object").columns.tolist())
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.decomposition import PCA

# Drop class label 'Diagnosis' if present
X_cluster = encoded_df.drop(columns=['Diagnosis', 'PatientID' , 'DoctorInCharge'], errors='ignore')
X_cluster = X_cluster.select_dtypes(include='number')
X_scaled = MinMaxScaler().fit_transform(X_cluster)
print(X_scaled.shape)
```

```python
k_range = range(2, 11)
wss_list, sil_list = [], []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    wss_list.append(km.inertia_)
    sil_list.append(silhouette_score(X_scaled, labels))

fig, ax = plt.subplots(1, 2, figsize=(12, 4))

ax[0].plot(list(k_range), wss_list, marker='o')
ax[0].set_title('Elbow Method (Total within-cluster SS)')
ax[0].set_xlabel('K'); ax[0].set_ylabel('WSS')

ax[1].plot(list(k_range), sil_list, marker='o', color='green')
ax[1].set_title('Average Silhouette Width')
ax[1].set_xlabel('K'); ax[1].set_ylabel('Silhouette')

for a in ax:
    for k in (2, 3, 4):
        a.axvline(k, color='red', linestyle='--', alpha=0.3)

plt.tight_layout(); plt.show()
```

```python
chosen_k = [2, 3, 4]
models, results = {}, []

for k in chosen_k:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    models[k] = (km, labels)
    results.append({
        'K': k,
        ' Silhouette ': silhouette_score(X_scaled, labels),
        'Total within-cluster SS (WSS)': km.inertia_,
        'Davies-Bouldin (lower=better)': davies_bouldin_score(X_scaled, labels),
        'Cluster sizes': sorted(np.bincount(labels).tolist(), reverse=True)
    })

results_df = pd.DataFrame(results).set_index('K')
display(results_df.round(4))
```

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
print('Variance explained by PC1 + PC2: {:.1f}%'.format(pca.explained_variance_ratio_.sum() * 100))

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, k in zip(axes, chosen_k):
    km, labels = models[k]
    centers_pca = pca.transform(km.cluster_centers_)
    ax.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='tab10', s=15, alpha=0.6)
    ax.scatter(centers_pca[:, 0], centers_pca[:, 1], c='black', marker='X',
               s=200, edgecolors='white', label='Centroids')
    ax.set_title(f'K = {k}')
    ax.set_xlabel('PC1'); ax.set_ylabel('PC2'); ax.legend()
plt.suptitle('K-means clusters projected on the first two principal components')
plt.tight_layout(); plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for ax, k in zip(axes, chosen_k):
    _, labels = models[k]
    sil_vals = silhouette_samples(X_scaled, labels)
    y_lower = 10
    for c in range(k):
        v = np.sort(sil_vals[labels == c])
        y_upper = y_lower + len(v)
        ax.fill_betweenx(np.arange(y_lower, y_upper), 0, v, alpha=0.7, label=f'Cluster {c}')
        y_lower = y_upper + 10
    ax.axvline(sil_vals.mean(), color='red', linestyle='--', label='Average')
    ax.set_title(f'K = {k}  (avg = {sil_vals.mean():.4f})')
    ax.set_xlabel('Silhouette coefficient'); ax.set_yticks([]); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()
```

```python
ks = np.array(list(k_range), dtype=float)
w = np.array(wss_list)
xn = (ks - ks.min()) / (ks.max() - ks.min())
yn = (w - w.min()) / (w.max() - w.min())
knee_dist = dict(zip(k_range, np.abs(yn - (1 - xn)) / np.sqrt(2)))

k_values = chosen_k
sil = results_df[' Silhouette '].to_dict()
dbi = results_df['Davies-Bouldin (lower=better)'].to_dict()

votes = {
    'Silhouette (max)': max(k_values, key=lambda k: sil[k]),
    'Elbow / WSS (knee)': max(k_values, key=lambda k: knee_dist[k]),
    'Davies-Bouldin (min)': min(k_values, key=lambda k: dbi[k]),
}
for name, v in votes.items():
    print(f"{name:22s} -> K = {v}")

best_k = Counter(votes.values()).most_common(1)[0][0]
print("Best K by majority rule: K =", best_k)
```

## Clustering

We applied K-means to explore groups of similar patient records in the Diabetes Health Dataset [1]. We used the 44 preprocessed features and excluded Diagnosis when fitting the models. After clustering, we compared the diagnosis percentages to help describe the groups.

### Comparing the clustering results

The following table summarizes the results of the three K values tested in Section [6]. We compare these results before discussing the characteristics of the selected clusters.

```python
import pandas as pd

from IPython.display import display, Markdown
from sklearn.metrics import silhouette_score, davies_bouldin_score

# Use the fitted models from Section [6].
comparison_rows = []

for k in [2, 3, 4]:
    model, labels = models[k]

    comparison_rows.append({
        "K": k,
        "Silhouette": silhouette_score(X_scaled, labels),
        "WSS": model.inertia_,
        "Davies–Bouldin": davies_bouldin_score(X_scaled, labels)
    })

comparison = pd.DataFrame(comparison_rows).set_index("K")
display(comparison.round(4))

k2 = comparison.loc[2]
k3 = comparison.loc[3]
k4 = comparison.loc[4]

display(Markdown(
    f"**K = 2.** The silhouette was {k2['Silhouette']:.4f} "
    f"and WSS was {k2['WSS']:.2f}. Two groups gave a simple "
    "partition, but the low silhouette suggests weak separation "
    "between the groups.\n\n"

    f"**K = 3.** The silhouette was {k3['Silhouette']:.4f}, "
    f"with WSS {k3['WSS']:.2f} and Davies–Bouldin "
    f"{k3['Davies–Bouldin']:.4f}. Davies–Bouldin preferred "
    "three clusters, while silhouette gave it the lowest score "
    "among the three choices. The measures did not fully agree.\n\n"

    f"**K = 4.** The silhouette was {k4['Silhouette']:.4f} "
    f"and WSS was {k4['WSS']:.2f}. Section [6] selected four "
    "clusters because silhouette and the elbow criterion favored "
    "it, giving it two of the three votes. It also had the lowest "
    "WSS among the tested choices. However, WSS normally decreases "
    "as K increases, so we considered the other measures as well."
))
```

We selected K = 4 because both the elbow criterion and the silhouette score favored four clusters, while the Davies–Bouldin index favored three. The silhouette scores for K = 2 and K = 4 were very close (0.0467 and 0.0468), so the improvement in separation was small.

### Characteristics of the selected clusters

We calculated the number of records and the percentage of each characteristic within every cluster. Age = 2 represents the Senior category, which includes patients aged 60 or above. BMI = 3 represents the Obese category, which includes BMI values of 30 or above. The two blood-sugar columns use category 2 for the diabetes range defined during preprocessing.

Gender codes are shown as recorded in the supplied dataset. Each percentage is calculated using the number of records in that cluster as the denominator.

```python
# Keep the same row order used by K-means.
profile = X_cluster.copy()
profile["Cluster"] = models[best_k][1]

# Add Diagnosis for interpretation after clustering.
profile["Diagnosis"] = encoded_df.loc[
    profile.index, "Diagnosis"
]

indicators = pd.DataFrame({
    "Gender = 1": profile["Gender"],
    "Statins = 1": profile["Statins"],
    "Senior": profile["Age"].eq(2),
    "Obese": profile["BMI"].eq(3),
    "Family history": profile["FamilyHistoryDiabetes"],
    "Hypertension": profile["Hypertension"],
    "FBS category 2": profile["FastingBloodSugar"].eq(2),
    "HbA1c category 2": profile["HbA1c"].eq(2),
    "Diagnosis = 1": profile["Diagnosis"]
})

rates = (
    indicators.groupby(profile["Cluster"])
    .mean()
    .mul(100)
)

counts = profile.groupby("Cluster").size()

summary = rates.round(1).rename(
    columns=lambda name: name + " (%)"
)
summary.insert(0, "Records", counts)

display(summary)

# Describe each cluster using its calculated percentages.
for cluster_id, row in rates.iterrows():
    display(Markdown(
        f"**Cluster {cluster_id} "
        f"({counts.loc[cluster_id]} records).** "
        f"Gender = 1 accounts for {row['Gender = 1']:.1f}% "
        f"of this group, and {row['Statins = 1']:.1f}% "
        "have Statins = 1. "
        f"Senior patients make up {row['Senior']:.1f}%, "
        f"while {row['Obese']:.1f}% are in the obese category. "
        "A family history of diabetes is recorded for "
        f"{row['Family history']:.1f}%, and hypertension "
        f"for {row['Hypertension']:.1f}%. "
        "The proportion with Diagnosis = 1 is "
        f"{row['Diagnosis = 1']:.1f}%."
    ))
```

### What we learned

From the clustering results, we noticed that the four groups mainly differed in Gender and Statins. Each cluster represented a different combination of these two features. Other characteristics, such as age, obesity, family history and hypertension, showed smaller differences between the groups.

Cluster 3 had the highest percentages of senior patients and patients in the obese category, at 45.0% and 43.8%, respectively. Cluster 0 had the highest percentage of patients with a family history of diabetes, at 26.4%. Cluster 2 had the highest percentage of patients with hypertension, at 16.0%.

The percentage of patients with Diagnosis = 1 ranged from 38.3% to 42.1%. This difference was relatively small, and all four clusters included patients with and without a diabetes diagnosis. Therefore, we could not clearly describe the clusters as low-, medium- or high-risk groups. The low silhouette scores also showed that the clusters were not strongly separated when all features were considered.

Clustering helped us understand which characteristics were shared by patients in each group. However, it did not separate the patients into groups with very different diagnosis rates. These findings depend on the features and preprocessing methods used. Also, the results do not show that statin use caused any difference in diabetes status.


### Comparison with the research paper

Kaliappan et al. [2] studied diabetes prediction using different datasets and classification methods. In Section 5.5, they identified age, family history of diabetes, and high blood pressure as important features in Dataset 1. Age, hypertension, and BMI were also important features in Dataset 2.

The percentage of senior patients ranged from 41.6% to 45.0%, and obesity ranged from 38.9% to 43.8%. These differences were relatively small across the four clusters.

While the research paper highlighted these features as important for diabetes prediction, our clustering results showed that the groups mainly differed in Gender and Statins. This difference may be explained by the use of different datasets and data mining techniques. The research paper focused on classification and feature selection, while our study used unsupervised clustering to identify groups of similar patients.

Overall, our results suggest that K-means can help identify similarities between patients. The low silhouette score showed weak separation across the 44 features used for clustering. When we compared diabetes diagnoses across the four clusters, the percentages were similar, ranging from 38.3% to 42.1%. Therefore, clustering was useful for understanding patient characteristics, but it did not clearly identify different diabetes risk groups.
