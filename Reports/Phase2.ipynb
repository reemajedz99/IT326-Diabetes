# 2. Discretization


```python
import numpy as np
import pandas as pd

# Clinical threshold bins based on ADA, WHO, and ACC/AHA guidelines
clinical = {
    'Age': ([-np.inf, 40, 60, np.inf], ['Young adult', 'Middle-aged', 'Senior']),
    'BMI': ([-np.inf, 18.5, 25, 30, np.inf], ['Underweight', 'Normal', 'Overweight', 'Obese']),
    'HbA1c': ([-np.inf, 5.7, 6.5, np.inf], ['Normal', 'Prediabetes', 'Diabetes']),
    'FastingBloodSugar': ([-np.inf, 100, 126, np.inf], ['Normal', 'Prediabetes', 'Diabetes']),
    'SystolicBP': ([-np.inf, 120, 130, 140, np.inf], ['Normal', 'Elevated', 'Stage 1', 'Stage 2']),
    'DiastolicBP': ([-np.inf, 80, 90, np.inf], ['Normal', 'Stage 1', 'Stage 2'])
}

binned = pd.DataFrame({
    c: pd.cut(raw_df[c], bins=e, labels=l, right=False)
    for c, (e, l) in clinical.items()
})

# Verify no NaNs were generated during binning
assert binned.notna().all().all()

# Display Raw vs Discretized Snapshots
print("Raw sample:")
display(raw_df[list(clinical)].head())

print("\nDiscretized sample:")
display(binned.head())

print("\nBin Distributions:")
for c in clinical:
    print(c, binned[c].value_counts(sort=False).to_dict())

# Apply changes to Preprocessed DataFrame
preprocessed_df = raw_df.copy()
for col in clinical:
    preprocessed_df[col] = binned[col]
```

**Justification:** In our diabetes dataset, features like blood sugar, blood pressure, and BMI directly correlate with diagnostic stages. Grouping raw patient measurements into tiers (such as 'Normal', 'Prediabetes', and 'Diabetes') turns continuous values into interpretable groups. Improving data quality and making extracted patterns human-readable for medical analysis.

**Why was Discretizing applied:** Exact numerical health measurements can contain small variations and minor differences that make classification models overly complex. Discretizing these continuous values into domain-based categories ensures results strictly follow real known medical standards rather than arbitrary mathematical splits.  

**How was Discretizing applied:** Left-inclusive interval mapping (pd.cut(..., right=False)) was applied directly to unscaled values from raw_df using standard medical thresholds. After binning, value_counts() was used to verify that patient instances were cleanly distributed across risk categories without producing missing values.

**On which attributes:** Age, BMI, HbA1c, FastingBloodSugar, SystolicBP, and DiastolicBP.
