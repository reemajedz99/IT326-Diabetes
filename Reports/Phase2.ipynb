The IQR method was applied to all numerical attributes to detect potential outliers. The results showed zero outliers for all numerical attributes, indicating that no outlier handling is required.

```python

Q1 = df[numerical].quantile(0.25)
Q3 = df[numerical].quantile(0.75)
IQR = Q3 - Q1

outliers = ((df[numerical] < (Q1 - 1.5 * IQR)) |
(df[numerical] > (Q3 + 1.5 * IQR))).sum()

outliers
```

## 1. Normalization


#### **Justification:**
In our raw dataset, numerical features have very different ranges. Scaling the data ensures that all features are uniform, preventing large numbers from dominating distance calculations and improving overall model performance. After scaling, `df.describe()` was used to verify that all transformed features are bounded between `0` and `1`.


#### **Why was Normalization applied:**

Unscaled features can distort machine learning model performance, especially algorithms that rely on distance calculations. Normalizing continuous numerical attributes ensures that every variable contributes equally to model learning without altering the original data distribution or relationships.

#### **How was Normalization applied:**

MinMaxScale (MinMaxScaler from sklearn.preprocessing) was applied to continuous numerical features to transform them linearly into a range between 0 and 1. After scaling, df.describe() was used to verify that minimum and maximum bounds were correctly set to 0 and 1 across all selected features.


#### **On which attributes:**
`AlcoholConsumption`, `PhysicalActivity`, `DietQuality`, `SleepQuality`, `SerumCreatinine`, `BUNLevels`, `CholesterolTotal`, `CholesterolLDL`, `CholesterolHDL`, `CholesterolTriglycerides`, `FatigueLevels`, and `QualityOfLifeScore`.

#### **How did Normalization improve the dataset:**
 The resulting dataset provides balanced, uniformly scaled inputs that prevent feature bias, improving overall convergence and accuracy for subsequent machine learning models.




```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

preprocessed_df = df.copy()

num_cols = [
    'AlcoholConsumption', 'PhysicalActivity', 'DietQuality',
    'SleepQuality', 'SerumCreatinine', 'BUNLevels',
    'CholesterolTotal', 'CholesterolLDL', 'CholesterolHDL',
    'CholesterolTriglycerides', 'FatigueLevels', 'QualityOfLifeScore'
]

scaler = MinMaxScaler()
preprocessed_df[num_cols] = scaler.fit_transform(preprocessed_df[num_cols])

print("Raw Dataset Sample:")
display(df[num_cols].head())

print("\nNormalized Dataset Sample:")
display(preprocessed_df[num_cols].head())
```

## 2. Discretization


```python
import numpy as np
import pandas as pd

# Clinical threshold bins based on ADA, WHO, and ACC/AHA guidelines
clinical = {
    'Age': ([-np.inf, 40, 60, np.inf], ['Young adult', 'Middle-aged', 'Senior']),
    'BMI': ([-np.inf, 18.5, 25, 30, np.inf], ['Underweight', 'Normal', 'Overweight', 'Obese']),
    'HbA1c': ([-np.inf, 5.7, 6.5, np.inf], ['Normal', 'Prediabetes', 'Diabetes']),
    'FastingBloodSugar': ([-np.inf, 100, 126, np.inf], ['Normal', 'Prediabetes', 'Diabetes']),
    'SystolicBP': ([-np.inf, 120, 130, 140, np.inf], ['Normal', 'Elevated', 'Stage 1', 'Stage 2']),
    'DiastolicBP': ([-np.inf, 80, 90, np.inf], ['Normal', 'Stage 1', 'Stage 2'])
}

binned = pd.DataFrame({
    c: pd.cut(df[c], bins=e, labels=l, right=False)
    for c, (e, l) in clinical.items()
})

# Verify no NaNs were generated during binning
assert binned.notna().all().all()

# Display Raw vs Discretized Snapshots
print("Raw sample:")
display(df[list(clinical)].head())

print("\nDiscretized sample:")
display(binned.head())

print("\nBin Distributions:")
for c in clinical:
    print(c, binned[c].value_counts(sort=False).to_dict())

# Apply changes to Preprocessed DataFrame
for col in clinical:
    preprocessed_df[col] = binned[col]
```

**Justification:**

In our diabetes dataset, features like blood sugar, blood pressure, and BMI directly correlate with diagnostic stages. Grouping raw patient measurements into tiers (such as 'Normal', 'Prediabetes', and 'Diabetes') turns continuous values into interpretable groups. Improving data quality and making extracted patterns human-readable for medical analysis.

**Why was Discretizing applied:**

Exact numerical health measurements can contain small variations and minor differences that make classification models overly complex. Discretizing these continuous values into domain-based categories ensures results strictly follow real known medical standards rather than arbitrary mathematical splits.  

**How was Discretizing applied:**

Left-inclusive interval mapping (pd.cut(..., right=False)) was applied directly to unscaled values from raw_df using standard medical thresholds. After binning, value_counts() was used to verify that patient instances were cleanly distributed across risk categories without producing missing values.

**On which attributes:**

Age, BMI, HbA1c, FastingBloodSugar, SystolicBP, and DiastolicBP.

**How did Discretizing improve the dataset:**
The resulting dataset is more interpretable, as continuous clinical measurements are represented as meaningful categories that make patient patterns and risk levels easier to identify.

# 3. Variable Transformation

```python

df["Cholesterol_Ratio"] = df["CholesterolTotal"] / df["CholesterolHDL"]

print("Raw dataset snapshot:")
print(df[["PatientID", "CholesterolTotal", "CholesterolHDL", "Diagnosis"]].head(10))

print("\nPreprocessed dataset snapshot:")
print(df[["PatientID", "CholesterolTotal", "CholesterolHDL","Cholesterol_Ratio", "Diagnosis"]].head(10).round(3))

print("\nCholesterol_Ratio summary:")
print(df["Cholesterol_Ratio"].describe().round(2))
```

The code applies variable transformation by deriving a new attribute, Cholesterol_Ratio, from two existing attributes: CholesterolTotal divided by CholesterolHDL. The original columns were kept unchanged.

**Justification:**

Total cholesterol and HDL each describe only part of a patient's lipid profile. Their ratio is a widely used clinical indicator of metabolic and cardiovascular risk, and it summarizes both measurements in a single meaningful value. Creating it gives the mining algorithms a more informative feature than either raw attribute alone.

**Preprocessing Result:**

A new numeric attribute, Cholesterol_Ratio, was added (range 1.57–14.41, mean 4.55),captures CholesterolTotal divided and CholesterolHDL balance in a single value.

**How it improved the dataset:**

The transformation combined two related measurements into one interpretable feature that captures the balance between total and "good" cholesterol, enriching the dataset for classification and clustering without losing any original information.
