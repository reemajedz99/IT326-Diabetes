## Outlier Detection
The IQR method was applied to all numerical attributes to detect potential outliers. The results showed zero outliers for all numerical attributes, indicating that no outlier handling is required.

```python

Q1 = df[numerical].quantile(0.25)
Q3 = df[numerical].quantile(0.75)
IQR = Q3 - Q1

outliers = ((df[numerical] < (Q1 - 1.5 * IQR)) |
(df[numerical] > (Q3 + 1.5 * IQR))).sum()

outliers
```

## Box Plot
the box plot shows the distribution of BMI values, including the median, quartiles, and the range of data. No values appear outside the whiskers, indicating that there are no outliers in the BMI attribute. Therefore, no preprocessing for outlier handling is needed for BMI.

```python
import matplotlib.pyplot as plt
plt.boxplot(df['BMI'])
plt.title('Box plot of BMI')
plt.ylabel('BMI')
plt.show()
```

# Data Preprocessing

### 1. Normalization

The code normalizes the data by applying `MinMaxScaler` from `sklearn.preprocessing`. It scales all 18 continuous numeric columns in the dataset so that their values are transformed into a range between `0` and `1`, while keeping the target column (`Diagnosis`) unchanged.

#### **Applied Attributes:**
`Age`, `BMI`, `AlcoholConsumption`, `PhysicalActivity`, `DietQuality`, `SleepQuality`, `SystolicBP`, `DiastolicBP`, `FastingBloodSugar`, `HbA1c`, `SerumCreatinine`, `BUNLevels`, `CholesterolTotal`, `CholesterolLDL`, `CholesterolHDL`, `CholesterolTriglycerides`, `FatigueLevels`, and `QualityOfLifeScore`.

#### **Justification:**
In our raw dataset, numerical features have very different ranges. Scaling the data ensures that all features are uniform, preventing large numbers from dominating distance calculations and improving overall model performance. After scaling, `df.describe()` was used to verify that all transformed features are bounded between `0` and `1`.


```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
import os

raw_df = pd.read_csv('Dataset/Raw_dataset.csv')

preprocessed_df = raw_df.copy()

num_cols = [
    'Age', 'BMI', 'AlcoholConsumption', 'PhysicalActivity',
    'DietQuality', 'SleepQuality', 'SystolicBP', 'DiastolicBP',
    'FastingBloodSugar', 'HbA1c', 'SerumCreatinine', 'BUNLevels',
    'CholesterolTotal', 'CholesterolLDL', 'CholesterolHDL',
    'CholesterolTriglycerides', 'FatigueLevels', 'QualityOfLifeScore'
]


scaler = MinMaxScaler()
preprocessed_df[num_cols] = scaler.fit_transform(preprocessed_df[num_cols])

print(" Raw Dataset Sample ")
display(raw_df[num_cols].head())

print("\nNormalized Dataset Sample ")
display(preprocessed_df[num_cols].head())

if not os.path.exists('Dataset'):
    os.makedirs('Dataset')

preprocessed_df.to_csv('Dataset/Preprocessed_dataset.csv', index=False)

import os

if not os.path.exists('Dataset'):
    os.makedirs('Dataset')

preprocessed_df.to_csv('Dataset/Preprocessed_dataset.csv', index=False)

```
